# hyprp on Kaggle

GPU Fermat PRP test of N = 2^n - C (base 3, same residue as Prime95 PRP-3 residue-type 1). hyprp computes each FFT in FP32 and exactly modulo 2^31-1 at once; it runs through CUDA on NVIDIA GPUs and OpenCL elsewhere. On one T4 a 3.6M-bit exponent takes about 7 minutes.

Settings (right panel): **Accelerator = GPU T4 x2**, **Internet = On**.

1. `MODE = 'selftest'`: build, check n=20011 and four larger n against known residues (on CUDA, the backend hyprp uses on NVIDIA), benchmark 20000 iterations at n=4,000,037 on each GPU (a few minutes).
2. `MODE = 'verify'`: full tests of exponents Prime95 already did (one per GPU, about 8 minutes), compared with Prime95's res64.
3. `MODE = 'bench'`: speed of every FFT length 16K..2048K on each GPU, one GPU at a time (the T4-Hour benchmark: 1 T4-Hour = what one T4 card computes in one hour); JSON results in the output.
4. `MODE = 'run'`: test `EXPONENTS` (and/or an input file `exponents.txt`, one n per line). Use **Save Version -> Save & Run All (Commit)**; results go to `results.jsonl`, checkpoints to `ckpt/` in the output. Next session: **Add Input -> Your Work -> this notebook** (latest output) and run again: finished exponents are skipped, unfinished ones resume from their checkpoints.

Progress is printed every minute; the full log of each exponent is `ckpt/hyprp_<n>.log`.

```python
MODE = 'selftest'          # 'selftest' | 'verify' | 'bench' | 'run'
C = 25                     # N = 2^n - C
EXPONENTS = ''             # for MODE = 'run', e.g. '3600029,3600061'
HOURS = 11.5               # Kaggle sessions end at 12 h; unfinished tests keep their checkpoint
REPO = 'https://github.com/RP-12000/dfprp.git'
BRANCH = 'master'
```
