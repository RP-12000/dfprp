# dfprp on Kaggle

GPU Fermat PRP test of N = 2^n - C (base 3, same residue as Prime95 PRP-3 residue-type 1), double-float FFT in OpenCL.

Settings (right panel): **Accelerator = GPU T4 x2** (or GPU P100), **Internet = On**.

1. `MODE = 'selftest'`: build, check n=20011, benchmark 20000 iterations at n=4,000,037 on each GPU (about 1 minute); prints minutes per 4M-bit exponent.
2. `MODE = 'verify'`: full tests of exponents Prime95 already did (one per GPU), compared with Prime95's res64.
3. `MODE = 'run'`: test `EXPONENTS` (and/or an input file `exponents.txt`). **Save Version -> Save & Run All (Commit)**; results go to `results.jsonl`, checkpoints to `ckpt/` in the output. Next session: **Add Input -> Your Work -> this notebook** (latest output) and run again: finished exponents are skipped, unfinished ones resume.

```python
MODE = 'selftest'          # 'selftest' | 'verify' | 'run'
C = 25                     # N = 2^n - C
EXPONENTS = ''             # for MODE = 'run', e.g. '3600029,3600061'
HOURS = 11.5               # Kaggle sessions end at 12 h; unfinished tests keep their checkpoint
REPO = 'https://github.com/RP-12000/dfprp.git'
```
